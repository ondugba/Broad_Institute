```python
strip_dupes['created_at'] = pd.to_datetime(strip_dupes['created_at'])
date_filter = pd.Timestamp('2024-08-13').date()
```

```python
df_removed_date = strip_dupes[strip_dupes['created_at'].dt.date != date_filter]
df_filtered = strip_dupes[strip_dupes['created_at'].dt.date == date_filter]
```

```python
df_unique = df_filtered.drop_duplicates(subset='sample_reference')
df_unique
```

```python
str_filtered = pd.concat([df_removed_date, df_unique], ignore_index=True)
str_filtered
```
