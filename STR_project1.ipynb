```python
def compare_dataframes_on_column(seq, merged_df, seq_screen_entry):
    seq_set = set(seq[seq_screen_entry])
    merged_set = set(merged_df[seq_screen_entry])
    comparison_result = {
        'seq_screen_entry': [],
        'identical': []
    }
    all_entries = seq_set.union(merged_set)
    for entry in all_entries:
        comparison_result['seq_screen_entry'].append(entry)
        comparison_result['identical'].append(entry in seq_set and entry in merged_set)

    result_df = pd.DataFrame(comparison_result)
    return result_df
result1 = compare_dataframes_on_column(seq, merged_df, 'seq_screen_entry')
result1
```

```python
def compare_dataframes_on_index(seq, merged_df):
    combined_indices = set(seq.index).union(set(merged_df.index))
    comparison_results = {
        'index': [],
        'identical': []
    }
    for idx in combined_indices:
        comparison_results['index'].append(idx)
        comparison_results['identical'].append(idx in seq.index and idx in merged_df.index)
    result_df = pd.DataFrame(comparison_results)  
    return result_df
result3 = compare_dataframes_on_index(seq, merged_df)
result3
```
