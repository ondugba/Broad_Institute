```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
#importing data from saved and editing csv file AFTER values
all_str = pd.read_csv('all_str_single_upload.csv')
all_str
```

```python
#BEFORE values
str_before = pd.read_csv('file_2.csv')
str_before
```

```python
#after
all_str['is_reference'].value_counts()
```

```python
#before
str_before['is_reference'].value_counts()
```

```python
#values for after
after_true = all_str['is_reference'].sum()
after_false = (~all_str['is_reference']).sum()
after_total = after_true + after_false
```

```python
#values for before
before_true = str_before['is_reference'].sum()
before_false = (~str_before['is_reference']).sum()
before_total = before_true + before_false
```

```python
#GPP values after
total_gpp_strs_after = all_str[all_str['source_group'] == 'GPP'].shape[0]
total_gpp_strs_after
```

```python
#GPP values before
total_gpp_strs_before = str_before[str_before['source_group'] == 'GPP'].shape[0]
total_gpp_strs_before
```

```python
#after
df_under75_after = all_str[all_str['percentage_match_to_parental'] < 0.75].shape[0]
df_under75_after
```

```python
#before
df_under75_before = str_before[str_before['percentage_match_to_parental'] < 0.75].shape[0]
df_under75_before
```

```python
#After
df_above75_after = all_str[all_str['percentage_match_to_parental'] >= 0.75].shape[0]
df_above75_after
```

```python
#before
df_above75_before = str_before[str_before['percentage_match_to_parental'] >= 0.75].shape[0]
df_above75_before
```

```python
categories = ['Total STRs', 'Total Reference STRs']
before_values = [before_total, before_true]
after_values = [after_total, after_true]

# Create the bar chart
bar_width = 0.35
index = np.arange(len(categories))

fig, ax = plt.subplots()
bar1 = ax.bar(index, before_values, bar_width, label='Before')
bar2 = ax.bar(index + bar_width, after_values, bar_width, label='After')

# Add labels, title, and ticks
ax.set_xlabel('Categories')
ax.set_ylabel('Count')
ax.set_title('Before and After Comparison of STRs')
ax.set_xticks(index + bar_width / 2)
ax.set_xticklabels(categories, rotation=45, ha='right')
ax.legend()

# Add annotations for the 'Before' bars
for i, value in enumerate(before_values):
    ax.text(index[i] - bar_width / 2, value + 0.02 * max(before_values), str(value), 
            ha='center', va='bottom')

# Add annotations for the 'After' bars
for i, value in enumerate(after_values):
    ax.text(index[i] + bar_width + bar_width / 2, value + 0.02 * max(after_values), str(value), 
            ha='center', va='bottom')

# Show the plot
plt.tight_layout()
plt.show()
```

```python


# Data for the graphs
# Graph 1
total_strs = [before_total, after_total]

# Graph 2
total_reference_strs = [before_true, after_true]

# Graph 3

match_percentage_over = [df_above75_before, df_above75_after]
match_percentage_under = [df_under75_before, df_under75_after]

# Graph 4
#total_gpp_strs = [40, 4943]
total_gpp_strs = [total_gpp_strs_before, total_gpp_strs_after]

# Create figure and axes
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# Graph 1
axs[0, 0].bar(['Before', 'After'], total_strs, color=['blue', 'green'])
axs[0, 0].set_title('Total STRs Uploaded')
axs[0, 0].set_ylabel('Count')

# Graph 2
axs[0, 1].bar(['Before', 'After'], total_reference_strs, color=['blue', 'green'])
axs[0, 1].set_title('Total Reference STRs')
axs[0, 1].set_ylabel('Count')

# Graph 3
bar_width = 0.35
index = range(2)

# STRs with percentage match >=0.75
axs[1, 0].bar(index, match_percentage_over, bar_width, label='Match >= 0.75', color='blue')
# STRs with percentage match <0.75
axs[1, 0].bar([i + bar_width for i in index], match_percentage_under, bar_width, label='Match < 0.75', color='red')
axs[1, 0].set_title('STRs with Percentage Match')
axs[1, 0].set_xticks([i + bar_width / 2 for i in index])
axs[1, 0].set_xticklabels(['Before', 'After'])
axs[1, 0].set_ylabel('Count')
axs[1, 0].legend()

# Graph 4
axs[1, 1].bar(['Before', 'After'], total_gpp_strs, color=['blue', 'green'])
axs[1, 1].set_title('Total GPP STRs')
axs[1, 1].set_ylabel('Count')

# Adjust layout
plt.tight_layout()

# Show plots
plt.show()
```
